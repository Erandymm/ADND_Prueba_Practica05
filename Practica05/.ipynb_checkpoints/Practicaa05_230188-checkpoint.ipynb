## Data Science for Good: DonorsChoose

**Ciencia de Datos para el Bien: DonorsChoose**

DonorsChoose es una organización sin fines de lucro con sede en Estados Unidos que permite a cualquier persona donar directamente a proyectos de aulas de escuelas públicas.
```python
def bar_hor(df, col, title, color, w=None, h=None, lm=0, limit=100, return_trace=False, rev=False):
    cnt_srs = df[col].value_counts()
    yy = cnt_srs.head(limit).index[::-1] 
    xx = cnt_srs.head(limit).values[::-1] 
    if rev:
        yy = cnt_srs.tail(limit).index[::-1] 
        xx = cnt_srs.tail(limit).values[::-1] 
        
    
    trace = go.Bar(y=yy, x=xx, orientation = 'h', marker=dict(color=color))
    if return_trace:
        return trace 
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)

def bar_hor_noagg(x, y, title, color, w=None, h=None, lm=0, limit=100, rt=False):
    trace = go.Bar(y=x, x=y, orientation = 'h', marker=dict(color=color))
    if rt:
        return trace
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)


def bar_ver_noagg(x, y, title, color, w=None, h=None, lm=0, rt = False):
    trace = go.Bar(y=y, x=x, marker=dict(color=color))
    if rt:
        return trace
    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    data = [trace]
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)

    
trace1 = bar_hor(donors_df, 'Donor City', "Top Cities with maximum Donors", '#c2d2ed', 500, 400, 200, limit = 10, return_trace=True)
trace2 = bar_hor(donors_df, 'Donor State', "Top States with maximum Donors", '#c2d2ed', 500, 400, 200, limit = 10, return_trace=True)
trace4 = bar_hor(donors_df, 'Donor State', "Top States with maximum Donors", '#c2d2ed', 600, 400, 200, limit = 10, return_trace=True, rev=True)


fig = tools.make_subplots(rows=1, cols=2, print_grid=False, subplot_titles = ['States with maximum Donors','States with minimum Donors'])
# fig = tools.make_subplots(rows=2, cols=2, specs=[[{}, {}], [{'colspan': 2}, None]], print_grid=False, subplot_titles = ['States with maximum Donors','States with minimum Donors', 'Cities with maximum Donors'])
fig.append_trace(trace2, 1, 1);
fig.append_trace(trace4, 1, 2);
# fig.append_trace(trace1, 2, 1);

fig['layout'].update(height=400, showlegend=False);
iplot(fig); 
```

### Conjunto de Datos Ajustado por Población

Los estados que aparecen en las primeras posiciones de los gráficos anteriores corresponden, en su mayoría, a aquellos con una población muy elevada. Por ello, resulta más representativo analizar el **número de donantes por cada 100,000 habitantes**, ya que este indicador permite realizar comparaciones más justas entre estados con diferentes tamaños de población.

Para este análisis se incorporó un conjunto de datos ajustado por población, el cual incluye la población estimada de cada estado de Estados Unidos correspondiente al año **2013**. Gracias a este ajuste es posible identificar qué estados presentan una mayor participación de donantes en relación con el número de habitantes y no únicamente por su tamaño poblacional.

### 2.3 Estados con la Mayor Proporción de Donantes respecto a su Población

En esta sección se compara el número de donantes de cada estado en relación con su población total. En lugar de considerar únicamente la cantidad de donantes, este indicador muestra cuántos donantes existen por cada **100,000 habitantes**, permitiendo realizar comparaciones más equitativas entre estados con diferentes tamaños de población.

```python
census_2013 = {'Mississippi': 2991207, 'Iowa': 3090416, 'Oklahoma': 3850568, 'Delaware': 925749, 'Minnesota': 5420380, 'Alaska': 735132, 'Illinois': 12882135, 'Arkansas': 2959373, 'New Mexico': 2085287, 'Indiana': 6570902, 'Maryland': 5928814, 'Louisiana': 4625470, 'Texas': 26448193, 'Wyoming': 582658, 'Arizona': 6626624, 'Wisconsin': 5742713, 'Michigan': 9895622, 'Kansas': 2893957, 'Utah': 2900872, 'Virginia': 8260405, 'Oregon': 3930065, 'Connecticut': 3596080, 'New York': 19651127, 'California': 38332521, 'Massachusetts': 6692824, 'West Virginia': 1854304, 'South Carolina': 4774839, 'New Hampshire': 1323459, 'Vermont': 626630, 'Georgia': 9992167, 'North Dakota': 723393, 'Pennsylvania': 12773801, 'Florida': 19552860, 'Hawaii': 1404054, 'Kentucky': 4395295, 'Rhode Island': 1051511, 'Nebraska': 1868516, 'Missouri': 6044171, 'Ohio': 11570808, 'Alabama': 4833722, 'South Dakota': 844877, 'Colorado': 5268367, 'Idaho': 1612136, 'New Jersey': 8899339, 'Washington': 6971406, 'North Carolina': 9848060, 'Tennessee': 6495978, 'Montana': 1015165, 'District of Columbia': 646449, 'Nevada': 2790136, 'Maine': 1328302}

donors_from_states = dict(donors_df['Donor State'].value_counts())

don_pop = {}
for state, don in donors_from_states.items():
    if state not in census_2013:
        continue
    don_pop[state]  = float(don)*100000/ census_2013[state]

import operator
don_pop = sorted(don_pop.items(), key=operator.itemgetter(1), reverse = True)
xx = [x[0] for x in (don_pop)][1:]
yy = [x[1] for x in (don_pop)][1:]

trace2 = go.Bar(
    x=xx,
    y=yy,
    name='Donors to Population',
    marker=dict(color='green'),
    opacity=0.3
)

data = [trace2]
layout = go.Layout(
    barmode='group',
    legend=dict(dict(x=-.1, y=1.2)),
    margin=dict(b=120),
    title = 'States with highest Donors to Population Ratio',
)

fig = go.Figure(data=data, layout=layout)
iplot(fig, filename='grouped-bar')
```

![Gráfica](graficas/img3.png)

- Un resultado interesante del análisis es que **Carolina del Sur**, **Massachusetts** y **Connecticut** presentan la mayor proporción de donantes en relación con su población. Estos estados cuentan con aproximadamente **985**, **907** y **878 donantes por cada 100,000 habitantes**, respectivamente.

- En el extremo opuesto, **Wyoming**, **Nebraska** y **Nuevo México** registran la menor proporción de donantes, con alrededor de **285**, **335** y **363 donantes por cada 100,000 habitantes**, respectivamente.

- Aunque **Texas** se encuentra entre los estados con el mayor número total de donantes, su proporción de donantes respecto a la población es relativamente menor, con aproximadamente **508 donantes por cada 100,000 habitantes**. Esto demuestra la importancia de ajustar el análisis por población para obtener comparaciones más representativas.

## 7.2 ¿Cuáles son las categorías, subcategorías y categorías de recursos más frecuentes en los proyectos publicados?

```python
# Create some date time features 
projects_df['Posted Date'] = pd.to_datetime(projects_df['Project Posted Date'])

projects_df['Posted Year'] = projects_df['Posted Date'].dt.year
projects_df['Posted Month'] = projects_df['Posted Date'].dt.month
projects_df['Posted Day'] = projects_df['Posted Date'].dt.weekday
projects_df['Posted Quarter'] = projects_df['Posted Date'].dt.quarter

projects_df['Funded Date'] = pd.to_datetime(projects_df['Project Fully Funded Date'])
projects_df['Funded Year'] = projects_df['Funded Date'].dt.year
projects_df['Funded Month'] = projects_df['Funded Date'].dt.month
projects_df['Funded Day'] = projects_df['Funded Date'].dt.weekday
projects_df['Funded Quarter'] = projects_df['Funded Date'].dt.quarter
projects_df['Funding Time'] = projects_df['Funded Date'] - projects_df['Posted Date']



t = projects_df['Project Subject Category Tree'].value_counts()
x = list(t.index)
y = list(t.values)
r = {}
for i,val in enumerate(x):
    for each in val.split(","):
        x1 = each.strip()
        if x1 not in r:
            r[x1] = y[i]
        r[x1] += y[i]
sorted_x = list(sorted(r.items(), key=operator.itemgetter(1), reverse = True))[:10]
x1 = [a[0] for a in sorted_x][::-1]
y1 = [a[1] for a in sorted_x][::-1]



t1 = projects_df['Project Subject Subcategory Tree'].value_counts()
x2 = list(t1.index)
y2 = list(t1.values)
r = {}
for i,val in enumerate(x2):
    for each in val.split(","):
        x1_ = each.strip()
        if x1_ not in r:
            r[x1_] = y2[i]
        r[x1_] += y2[i]        
x2 = r.keys()
y2 = r.values()
sorted_x = list(sorted(r.items(), key=operator.itemgetter(1), reverse = True))[:10]
x2 = [a[0] for a in sorted_x][::-1]
y2 = [a[1] for a in sorted_x][::-1]

t = projects_df['Project Resource Category'].value_counts()
x3 = list(t.index[::-1])[:10]
y3 = list(t.values[::-1])[:10]


trace1 = bar_hor_noagg(x1, y1, "", "#8cf2a9", 600, 400, 10, rt=True)
trace2 = bar_hor_noagg(x2, y2, "", "#8cf2a9", 600, 400, 10, rt=True)
trace3 = bar_hor_noagg(x3, y3, "", "#8cf2a9", 600, 400, 10, rt=True)

fig = tools.make_subplots(rows=1, cols=5, print_grid=False, subplot_titles = ['Project Categories' ,"", 'Project Subcategories', '', "Resource Categories"])
fig.append_trace(trace1, 1, 1);
fig.append_trace(trace2, 1, 3);
fig.append_trace(trace3, 1, 5);
fig['layout'].update(height=400, showlegend=False, margin=dict(l=120));
iplot(fig);
```

![Gráfica](graficas/img32.png)

La gran mayoría de los proyectos (más de 800 mil) pertenecen a la categoría principal de **Alfabetización y Lenguaje**. **Matemáticas y Ciencias** es la segunda categoría más destacada. En comparación, se ha publicado una cantidad considerablemente menor de proyectos relacionados con **Cuidado y Alimentación** desde 2013.

Entre las subcategorías, **Alfabetización, Matemáticas y Escritura** son las más comunes dentro de los proyectos publicados por los docentes, mientras que una menor cantidad de proyectos corresponde a las categorías de **Desarrollo Temprano y Salud**. Las principales categorías de recursos solicitados por los profesores son **Tecnología Educativa, Espacios de Lectura, Escritorios y Almacenamiento**.

### 7.6 ¿En qué meses se publicaron los proyectos?


```python
t = projects_df['Posted Month'].value_counts()


lObjectsALLcnts = list(t.values)

lObjectsALLlbls = list(t.index)
mapp1 = {1:'Jan',2:'Feb',3:'Mar',4:'Apr',5:'May',6:'Jun',7:'Jul',8:'Aug',9:'Sep',10:'Oct',11:'Nov',12:'Dec'}
lObjectsALLlbls = [mapp1[x] for x in lObjectsALLlbls]

iN = len(lObjectsALLcnts)
arrCnts = np.array(lObjectsALLcnts)

theta=np.arange(0, 2*np.pi, 2*np.pi/iN)
width = (2*np.pi)/iN *0.5
bottom = 50

fig = plt.figure(figsize=(8,8))
ax = fig.add_axes([0.2, 0.1, 1, 0.9], polar=True)
fig.suptitle('Project Posted Month   ', fontsize=16)
bars = ax.bar(theta, arrCnts, width=width, bottom=bottom, color='#f3a0f7')
plt.axis('off')

rotations = np.rad2deg(theta)
for x, bar, rotation, label in zip(theta, bars, rotations, lObjectsALLlbls):
    lab = ax.text(x,bottom+bar.get_height() , label, ha='left', va='center', rotation=rotation, rotation_mode="anchor",)   
plt.show()
```

![Gráfica](graficas/img36.png)

En el tercer trimestre (Q3), la mayor cantidad de proyectos fueron publicados durante el mes de septiembre, seguido por agosto. Por otro lado, junio fue el mes con la menor cantidad de proyectos publicados.

### 8.1 Limpieza de Texto

Se desarrolla una función auxiliar para normalizar el texto y eliminar elementos innecesarios o ruido. Esta función de limpieza de texto realiza la eliminación de palabras vacías (*stopwords*) y la eliminación de signos de puntuación, con el fin de preparar los datos para su posterior análisis.


```python
# Instalación de los paquetes necesarios
!pip install textblob nltk seaborn matplotlib pandas
```

```python
import string
import itertools
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import nltk
from nltk.corpus import stopwords
from textblob import TextBlob

# Descargar el recurso de stopwords de NLTK indispensable para tu función
nltk.download('stopwords')

# Cargar la lista de palabras vacías en inglés
stopwords_list = stopwords.words("english")

def _clean(txt):
    # Manejo de valores nulos o flotantes inesperados en el texto
    if not isinstance(txt, str):
        return ""
        
    # 1. Convertir a minúsculas
    txt = txt.lower()
    
    # 2. Eliminación de puntuación
    txt = ''.join(x for x in txt if x not in string.punctuation)
    
    # 3. Eliminación de stopwords
    clean_txt = ""
    for word in txt.split():
        if word in stopwords_list:
            continue
        clean_txt += " "
        clean_txt += word
        
    # 4. Eliminación de palabras de ruido específicas
    noise = ['students', 'my', 'title', 'would', 'every', 'will', 'many', 'also', 'donotremoveessaydivider']
    for ns in noise:
        clean_txt = clean_txt.replace(ns, "")
        
    return clean_txt

# --- PROCESAMIENTO DE DATOS ---
# (Asegúrate de tener cargados previamente 'projects_df' y 'schools_df')

# Combinar datasets
projsch_df = projects_df.merge(schools_df, on='School ID', how='inner')

# Filtrar por año 2018 y mes 4 (Abril)
april18_df = projsch_df[(projsch_df['Posted Year'] == 2018) & (projsch_df['Posted Month'] == 4)].copy()

# Aplicar la función de limpieza a las columnas de texto
april18_df['clean_essay'] = april18_df['Project Essay'].apply(_clean)
april18_df['clean_need'] = april18_df['Project Need Statement'].apply(_clean)

print("¡Procesamiento y limpieza de texto completados con éxito!")
```

### 8.2 Análisis de Bolsa de Palabras (*Bag of Words Analysis*)

El análisis de **Bag of Words** permite identificar las palabras y frases más utilizadas dentro de los ensayos escritos por los docentes. Esta técnica transforma los textos en una representación basada en la frecuencia de aparición de términos, facilitando la identificación de los temas y necesidades más comunes expresadas en los proyectos.

### 8.2.1 ¿Cuáles son las palabras y frases más utilizadas por los docentes en los ensayos?


```python
### Bag of Words 
from collections import Counter
def generate_ngrams(txt, N):
    grams = [txt[i:i+N] for i in range(len(txt)-N+1)]
    grams = [" ".join(b) for b in grams]
    return grams 

april18_df['unigrams'] = april18_df['clean_essay'].apply(lambda x : generate_ngrams(x.split(), 1))
april18_df['bigrams'] = april18_df['clean_essay'].apply(lambda x : generate_ngrams(x.split(), 2))
april18_df['trigrams'] = april18_df['clean_essay'].apply(lambda x : generate_ngrams(x.split(), 3))

all_unigrams = []
for each in april18_df['unigrams']:
    all_unigrams.extend(each)
t = Counter(all_unigrams).most_common(25)
x = [a[0] for a in t]
y = [a[1] for a in t]

all_bigrams = []
for each in april18_df['bigrams']:
    all_bigrams.extend(each)
t1 = Counter(all_bigrams).most_common(25)
x1 = [a[0] for a in t1]
y1 = [a[1] for a in t1]


all_trigrams = []
for each in april18_df['trigrams']:
    all_trigrams.extend(each)
t2 = Counter(all_trigrams).most_common(25)
x2 = [a[0] for a in t2]
y2 = [a[1] for a in t2]

sns.set_style("dark")
fig, axes = plt.subplots(nrows=1, ncols=5, squeeze=True, figsize = (12,10));

bar = sns.barplot(y=x, x=y, ax=axes[0], palette='GnBu_d', edgecolor="white");
bar.set(xlabel='', xticks=[]);
axes[0].set_title("Top Keywords used by Teachers");

fig.delaxes(axes[1]);

bar1 = sns.barplot(y=x1, x=y1, ax=axes[2], palette='GnBu_d', edgecolor="white");
bar1.set(xlabel='', xticks=[]);
axes[2].set_title("Top Bigrams used by Teachers");

fig.delaxes(axes[3]);

bar1 = sns.barplot(y=x2, x=y2, ax=axes[4], palette='GnBu_d', edgecolor="white");
bar1.set(xlabel='', xticks=[]);
axes[4].set_title("Top Trigrams used by Teachers");
```

![Gráfica](graficas/img45.png)

Las palabras clave más utilizadas en todos los ensayos, estados y tipos de escuelas son **"school"**, **"learning"**, **"classroom"** y **"reading"**. En cuanto a los **bigrams** (combinaciones de dos palabras), los principales enfoques se centran en frases como **"flexible seating"** y **"english language"**. ### 9.1 Proyectos expirados vs. proyectos financiados

En esta sección se exploran y analizan las características principales de los proyectos que fueron financiados exitosamente, comparándolas con aquellos proyectos que expiraron. Para este análisis se consideran los proyectos publicados durante el último año disponible, es decir, **2017**, con el objetivo de identificar diferencias relevantes entre ambos tipos de proyectos.


```python
import pandas as pd
import numpy as np
import plotly.graph_objs as go
from plotly import tools
from plotly.offline import iplot

# ==============================================================================
# 1. PREPARACIÓN DE DATOS (Filtros y Merges corregidos)
# ==============================================================================

# Filtrar proyectos por año y estatus
expired = projects_df[(projects_df['Posted Year'] == 2017) & (projects_df['Project Current Status'] == 'Expired')].copy()
funded = projects_df[(projects_df['Posted Year'] == 2017) & (projects_df['Project Current Status'] == 'Fully Funded')].copy()

# Unir con los datos de profesores
expired = expired.merge(teachers_df, on="Teacher ID")
funded = funded.merge(teachers_df, on="Teacher ID")

# ¡CORRECCIÓN CRÍTICA!: Unir con schools_df para inyectar 'School Metro Type' y 'School State'
expired = expired.merge(schools_df, on="School ID")
funded = funded.merge(schools_df, on="School ID")


# ==============================================================================
# 2. DEFINICIÓN DE LA FUNCIÓN PARA GENERAR TRAZAS
# ==============================================================================

def get_trace(col, func, flag=None, leg=False):
    t1 = expired[col].value_counts()
    t2 = funded[col].value_counts()

    if flag == 'num':
        t1 = expired[expired['NumProjects'] <= 20][col].value_counts()    
        t2 = funded[funded['NumProjects'] <= 20][col].value_counts() 
    
    trace1 = func(
        x=t1.index,
        y=[float(a)*100/sum(t1.values) for a in t1.values],
        name='Expired',
        marker=dict(color='#e82041'),
        showlegend=leg,
        opacity=0.9
    )
    trace2 = func(
        x=t2.index,
        y=[float(a)*100/sum(t2.values) for a in t2.values],
        name='Funded',
        marker=dict(color='#0de286'),
        opacity=0.9, 
        showlegend=leg
    )
    return trace1, trace2


# ==============================================================================
# 3. GENERACIÓN DE TRAZAS BÁSICAS
# ==============================================================================

tr1, tr2 = get_trace('Project Grade Level Category', go.Scatter)
tr3, tr4 = get_trace('School Metro Type', go.Scatter)
tr5, tr6 = get_trace('School State', go.Bar)
tr7, tr8 = get_trace('Project Resource Category', go.Bar)
```

### 9.2 Proyectos con baja vs. alta duración de financiamiento

En esta sección se comparan las características de los proyectos que fueron financiados exitosamente en un periodo corto, es decir, **menos de un día**, frente a aquellos proyectos que tardaron más tiempo en conseguir financiamiento, específicamente aquellos que tardaron **más de dos meses**.


```python
import pandas as pd
import numpy as np
import plotly.graph_objs as go
from plotly import tools
from plotly.offline import iplot

# ==============================================================================
# 1. FILTRADO Y FUSIÓN DE DATOS (Corrección del KeyError: 'School Metro Type')
# ==============================================================================

# Extraer el tiempo de fondeo y limpiar NaTs
projects_df['time'] = projects_df['Funding Time'].apply(lambda x : str(x).split()[0])
tempdf = projects_df[projects_df['time'] != "NaT"].copy()
tempdf['time'] = tempdf['time'].apply(int)
tempdf = tempdf[tempdf['Posted Year'] == 2017]

# Segmentar duraciones (Corta vs Larga)
df1 = tempdf[tempdf['time'] <= 1].copy() 
df2 = tempdf[tempdf['time'] >= 60].copy() 

# Unir con los datos de profesores (teachers_df)
df1 = df1.merge(teachers_df, on="Teacher ID")
df2 = df2.merge(teachers_df, on="Teacher ID")

# ¡SOLUCIÓN CRÍTICA!: Agregar el merge con schools_df para inyectar 'School Metro Type' y 'School State'
df1 = df1.merge(schools_df, on="School ID")
df2 = df2.merge(schools_df, on="School ID")


# ==============================================================================
# 2. FUNCIÓN DE GENERACIÓN DE TRAZAS
# ==============================================================================

def get_trace(col, func, flag=None, leg=False):
    t1 = df1[col].value_counts()
    t2 = df2[col].value_counts()

    if flag == 'num':
        t1 = df1[df1['NumProjects'] <= 20][col].value_counts()    
        t2 = df2[df2['NumProjects'] <= 20][col].value_counts() 
    
    trace1 = func(
        x=t1.index,
        y=[float(a)*100/sum(t1.values) for a in t1.values],
        name='Small Funding Duration',
        marker=dict(color='#0de286'),
        showlegend=leg,
        opacity=0.9
    )
    trace2 = func(
        x=t2.index,
        y=[float(a)*100/sum(t2.values) for a in t2.values],
        name='Large Funding Duration',
        marker=dict(color='#e82041'),
        opacity=0.9, 
        showlegend=leg
    )
    return trace1, trace2


# ==============================================================================
# 3. CREACIÓN DE TRAZAS BÁSICAS
# ==============================================================================

tr1, tr2 = get_trace('Project Grade Level Category', go.Scatter)
tr3, tr4 = get_trace('School Metro Type', go.Scatter)
tr5, tr6 = get_trace('School State', go.Bar)
tr7, tr8 = get_trace('Project Resource Category', go.Bar)
tr9, tr10 = get_trace('Posted Month', go.Bar, leg=True)


# ==============================================================================
# 4. CONTEO DE PALABRAS DE ENSAYOS (Word Count)
# ==============================================================================

df1['wc'] = df1['Project Essay'].apply(lambda x : len(str(x).split()))
df2['wc'] = df2['Project Essay'].apply(lambda x : len(str(x).split()))

bins_wc = [0, 50, 100, 150, 200, 250, 300, 350, 400, 450, 500, 10000]
order_wc = ['0-50', '50-100', '100-150', '150-200', '200-250', '250-300', '300-350', '350-400', '400-450', '450-500', '500-10000']

# --- Procesar df1 ---
df1['wc'] = pd.cut(df1['wc'], bins_wc)
t1_wc = dict(df1['wc'].value_counts())
labels_t1 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t1_wc.keys()]
vals_t1 = [float(a)*100/sum(t1_wc.values()) for a in list(t1_wc.values())]

d_t1 = {labels_t1[i]: vals_t1[i] for i in range(len(labels_t1))}
values_tr13 = [d_t1.get(v, 0) for v in order_wc]

tr13 = go.Bar(
    x=order_wc,
    y=values_tr13,
    name='Small Funding Duration',
    marker=dict(color='#0de286'),
    showlegend=False,
    opacity=0.9
)

# --- CORRECCIÓN CRÍTICA: Procesar df2 usando sus propias variables t2_wc ---
df2['wc'] = pd.cut(df2['wc'], bins_wc)
t2_wc = dict(df2['wc'].value_counts())
labels_t2 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t2_wc.keys()]
vals_t2 = [float(a)*100/sum(t2_wc.values()) for a in list(t2_wc.values())]

d_t2 = {labels_t2[i]: vals_t2[i] for i in range(len(labels_t2))}
values_tr14 = [d_t2.get(v, 0) for v in order_wc]

tr14 = go.Bar(
    x=order_wc,
    y=values_tr14,
    name='Large Funding Duration',
    marker=dict(color='#e82041'),
    showlegend=False,
    opacity=0.9
)


# ==============================================================================
# 5. ANÁLISIS DE COSTO DEL PROYECTO (Project Cost)
# ==============================================================================

bins_cost = [0, 100, 500, 1000, 1500, 2000, 2500, 3000, 70000]
order_cost = ['100-500', '500-1000', '1000-1500', '1500-2000', '2000-2500', '2500-3000', '3000-70000']

# --- Procesar df1 Costo ---
df1['binnedcost'] = pd.cut(df1['cost'], bins_cost)
t1_cost = dict(df1['binnedcost'].value_counts())
labels_c1 = [str(b).replace("]","").replace("(","").replace(", ","-") for b in t1_cost.keys()]
vals_c1 = [float(a)*100/sum(t1_cost.values()) for a in list(t1_cost.values())]

d_c1 = {labels_c1[i]: vals_c1[i] for i in range(len(labels_c1))}
values_tr15 = [d_c1.get(v, 0) for v in order_cost]
```

Con base en las gráficas y comparaciones realizadas, se pueden destacar las siguientes observaciones:

## Proyectos financiados por un solo donante

Un mayor porcentaje de los proyectos que recibieron apoyo de **un solo donante** presentaron las siguientes características:

## Proyectos financiados por múltiples donantes

Por otro lado, un mayor porcentaje de los proyectos que recibieron apoyo de **múltiples donantes** compartían las siguientes características:

- Correspondían a los niveles educativos **Grade 3–12**.
- Fueron publicados por escuelas ubicadas en **zonas suburbanas y rurales**.
- Presentaban un **costo superior a 500 USD**.
- Los ensayos de los docentes eran **más extensos**, utilizando un mayor número de palabras.
- Se relacionaban principalmente con las categorías de **Tecnología (Technology)** y **Materiales (Supplies)**.
